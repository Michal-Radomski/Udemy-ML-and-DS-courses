```python
import pandas as pd

# From a list
data = [10, 20, 30, 40, 50]
series = pd.Series(data)
print(series)
type(series)
```

```python
series_named = pd.Series([1, 3, 5], index=['x', 'y', 'z'])
print(series_named['y'])  # Outputs: 3
```

```python
grades = {'Math': 95, 'Science': 88, 'History': 92}
series_grades = pd.Series(grades)
print(series_grades)
```

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'London', 'Tokyo']
}
df = pd.DataFrame(data)
df  # Displays the table in Jupyter
```

```python
print(df.head())  # First 5 rows (or all if small)
print(df.describe())  # Summary stats for Age column
```

```python
print(df['Age'])  # Single column
print(df[df['Age'] > 28])  # Filtered rows
print(type(df))
```
