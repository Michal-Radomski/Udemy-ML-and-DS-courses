```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
# Generate sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create the plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='sin(x)', color='blue', linewidth=2)
plt.title('Simple Sine Wave Plot')
plt.xlabel('X values')
plt.ylabel('sin(X)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
# Sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.exp(-x/5) * np.cos(x)

# Set global style
plt.style.use('seaborn-v0_8-darkgrid')  # Modern dark theme

# Create subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Left plot: Styled line with fill
ax1.plot(x, y1, '-.', color='#FF6B6B', linewidth=2.5, label='sin(x)', alpha=0.8)
ax1.fill_between(x, y1, alpha=0.3, color='#FF6B6B')
ax1.set_title('Styled Sine Wave', fontsize=14, fontweight='bold')
ax1.set_xlabel('X Axis', fontsize=12)
ax1.set_ylabel('Value', fontsize=12)
ax1.legend(loc='upper right')
ax1.grid(True, linestyle='--', alpha=0.5)

# Annotate a point
ax1.annotate('Peak', xy=(np.pi/2, 1), xytext=(4, 0.8),
             arrowprops=dict(arrowstyle='->', color='white', lw=1.5),
             fontsize=10, color='white', fontweight='bold')

# Right plot: Multiple lines with markers
colors = ['#4ECDC4', '#45B7D1', '#F18F01']
for i, y in enumerate([y2, y3, y1*0.5]):
    ax2.plot(x, y, '-', color=colors[i], linewidth=2, 
             marker='o', markersize=4, markevery=10, label=f'Curve {i+1}')
ax2.set_title('Multi-Line Plot with Markers', fontsize=14, fontweight='bold')
ax2.legend(frameon=True, fancybox=True, shadow=True)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Sample data - sales by product
categories = ['Laptops', 'Phones', 'Tablets', 'Headphones', 'Monitors']
sales = [450, 320, 180, 250, 380]

# Create the bar chart
plt.figure(figsize=(10, 6))
bars = plt.bar(categories, sales, 
               color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#F18F01', '#96CEB4'],
               edgecolor='black', linewidth=1.2, alpha=0.8)

# Customize appearance
plt.title('Product Sales by Category (2026)', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Product Categories', fontsize=12)
plt.ylabel('Sales ($K)', fontsize=12)

# Add value labels on top of bars
for bar, value in zip(bars, sales):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5,
             f'${value}K', ha='center', va='bottom', fontweight='bold')

# Rotate x-axis labels for better readability
plt.xticks(rotation=15, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.3)

# Improve layout
plt.tight_layout()
plt.show()
```

```python
# Load built-in dataset
tips = sns.load_dataset("tips")

# Create grouped bar plot by day and smoker status
plt.figure(figsize=(10, 6))
sns.catplot(data=tips, kind="bar", x="day", y="total_bill", hue="smoker", 
            height=6, aspect=1.5, palette="Set2")

plt.title("Average Total Bill by Day and Smoker Status", fontsize=16, fontweight='bold', pad=20)
plt.xlabel("Day of Week", fontsize=12)
plt.ylabel("Average Total Bill ($)", fontsize=12)
plt.xticks(rotation=0)
plt.show()
```
